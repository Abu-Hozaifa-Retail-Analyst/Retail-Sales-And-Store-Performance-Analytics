```python
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
start_date = "2024-01-01"
end_date = "2025-12-31"
```

### Generate Every Date

```python
dates = pd.date_range(
    start=start_date,
    end=end_date,
    freq="D"
)
```

```python
len(dates)
```

### Create the DataFrame

```python
df_date = pd.DataFrame({
    "date": dates
})
```

```python
df_date.head()
```

### Add Year

```python
df_date["year"] = df_date["date"].dt.year
```

### Add Quarter

```python
df_date["quarter"] = (
    "Q" + df_date["date"].dt.quarter.astype(str)
)
```

### Add Month Number

```python
df_date["month"] = df_date["date"].dt.month
```

### Add Month Name

```python
df_date["month_name"] = (
    df_date["date"].dt.month_name()
)
```

### Add Week Number

```python
df_date["week"] = (
    df_date["date"].dt.isocalendar().week.astype(int)
)
```

### Add Day of Week

```python
df_date["day_of_week"] = (
    df_date["date"].dt.day_name()
)
```

### Add Day Number

```python
df_date["day"] = df_date["date"].dt.day
```

### Add Weekend Flag

```python
df_date["is_weekend"] = (
    df_date["date"].dt.dayofweek >= 4
)
```

```python
df_date["is_weekend"].value_counts()
```

### Add Month-Year

```python
df_date["month_year"] = (
    df_date["date"].dt.strftime("%Y-%m")
)
```

### Add Year-Month Sort Key

```python
df_date["year_month"] = (
    df_date["year"] * 100
    + df_date["month"]
)
```

### Add Quarter Number

```python
df_date["quarter_number"] = (
    df_date["date"].dt.quarter
)
```
