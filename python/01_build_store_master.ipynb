```python
import pandas as pd
import numpy as np
from pathlib import Path

np.random.seed(42)
```

### Define Our Stores

```python
stores = [
    ("GM001", "GulfMart Riyadh North", "Riyadh", "Central", "Hypermarket"),
    ("GM002", "GulfMart Riyadh South", "Riyadh", "Central", "Hypermarket"),
    ("GM003", "GulfMart Riyadh Central", "Riyadh", "Central", "Supermarket"),
    ("GM004", "GulfMart Riyadh East", "Riyadh", "Central", "Supermarket"),
    ("GM005", "GulfMart Riyadh West", "Riyadh", "Central", "Supermarket"),
    ("GM006", "GulfMart Riyadh Express 1", "Riyadh", "Central", "Express"),
    ("GM007", "GulfMart Riyadh Express 2", "Riyadh", "Central", "Express"),

    ("GM008", "GulfMart Jeddah North", "Jeddah", "Western", "Hypermarket"),
    ("GM009", "GulfMart Jeddah South", "Jeddah", "Western", "Hypermarket"),
    ("GM010", "GulfMart Jeddah Central", "Jeddah", "Western", "Supermarket"),
    ("GM011", "GulfMart Jeddah East", "Jeddah", "Western", "Supermarket"),
    ("GM012", "GulfMart Jeddah Express", "Jeddah", "Western", "Express"),

    ("GM013", "GulfMart Dammam North", "Dammam", "Eastern", "Hypermarket"),
    ("GM014", "GulfMart Dammam Central", "Dammam", "Eastern", "Supermarket"),
    ("GM015", "GulfMart Dammam Express", "Dammam", "Eastern", "Express"),

    ("GM016", "GulfMart Khobar Central", "Khobar", "Eastern", "Supermarket"),
    ("GM017", "GulfMart Khobar Express", "Khobar", "Eastern", "Express"),

    ("GM018", "GulfMart Makkah Central", "Makkah", "Western", "Supermarket"),
    ("GM019", "GulfMart Makkah Express", "Makkah", "Western", "Express"),
    ("GM020", "GulfMart Makkah Neighborhood", "Makkah", "Western", "Neighborhood"),

    ("GM021", "GulfMart Madinah Central", "Madinah", "Western", "Supermarket"),
    ("GM022", "GulfMart Madinah Neighborhood", "Madinah", "Western", "Neighborhood"),

    ("GM023", "GulfMart Abha Central", "Abha", "Southern", "Supermarket"),
    ("GM024", "GulfMart Abha Neighborhood", "Abha", "Southern", "Neighborhood"),

    ("GM025", "GulfMart Tabuk Central", "Tabuk", "Northern", "Express"),
]
```

### Create the DataFrame

```python
df_store = pd.DataFrame(
    stores,
    columns=[
        "store_id",
        "store_name",
        "city",
        "region",
        "store_type"
    ]
)
```

```python
df_store.head()
```

### Add Store Size

```python
size_ranges = {
    "Hypermarket": (25000, 40000),
    "Supermarket": (12000, 22000),
    "Express": (4000, 8000),
    "Neighborhood": (6000, 10000)
}
```

```python
df_store["store_size_sqft"] = df_store["store_type"].apply(
    lambda store_type: np.random.randint(
        size_ranges[store_type][0],
        size_ranges[store_type][1] + 1
    )
)
```

```python
df_store[
    ["store_id", "store_type", "store_size_sqft"]
].head(10)
```

### Add Opening Dates

```python
opening_dates = pd.date_range(
    start="2018-01-01",
    end="2024-12-31",
    freq="D"
)
```

```python
df_store["opening_date"] = np.random.choice(
    opening_dates,
    size=len(df_store),
    replace=False
)
```

```python
df_store["opening_date"] = pd.to_datetime(
    df_store["opening_date"]
)
```

```python
df_store[
    ["store_id", "store_name", "opening_date"]
].head(10)
```

### Add Annual Sales Target

```python
target_ranges = {
    "Hypermarket": (10_000_000, 15_000_000),
    "Supermarket": (6_000_000, 10_000_000),
    "Express": (2_500_000, 5_000_000),
    "Neighborhood": (3_000_000, 5_500_000)
}
```

```python
df_store["target_sales"] = df_store["store_type"].apply(
    lambda store_type: np.random.randint(
        target_ranges[store_type][0],
        target_ranges[store_type][1] + 1
    )
)
```

```python
df_store[
    ["store_id", "store_type", "target_sales"]
]
```

### Add Target Margin

```python
margin_targets = {
    "Hypermarket": 0.28,
    "Supermarket": 0.30,
    "Express": 0.32,
    "Neighborhood": 0.31
}
```

```python
df_store["target_margin"] = df_store["store_type"].map(
    margin_targets
)
```

```python
df_store[
    ["store_id", "store_type", "target_margin"]
].head(10)
```
