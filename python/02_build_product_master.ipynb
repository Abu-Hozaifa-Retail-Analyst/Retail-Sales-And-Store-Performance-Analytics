```python
import pandas as pd
import numpy as np
from pathlib import Path

np.random.seed(42)
```

### Categories and subcategories

```python
categories = {
    "Food & Beverages": [
        "Beverages",
        "Snacks",
        "Dairy",
        "Grocery",
        "Bakery"
    ],
    
    "Personal Care": [
        "Hair Care",
        "Oral Care",
        "Body Care",
        "Shaving"
    ],
    
    "Household": [
        "Cleaning",
        "Laundry",
        "Kitchen",
        "Paper Products"
    ],
    
    "Beauty": [
        "Skincare",
        "Makeup",
        "Fragrance",
        "Hair Styling"
    ],
    
    "Health & Wellness": [
        "Vitamins",
        "Fitness",
        "Personal Health",
        "Wellness"
    ],
    
    "Baby Care": [
        "Diapers",
        "Baby Food",
        "Baby Bath",
        "Baby Accessories"
    ],
    
    "Electronics": [
        "Mobile Accessories",
        "Small Electronics",
        "Cables & Chargers",
        "Home Electronics"
    ],
    
    "Fresh Food": [
        "Fruits",
        "Vegetables",
        "Meat",
        "Bakery Fresh"
    ]
}
```

### Brands

```python
brands = [
    "Gulf Fresh",
    "Desert Harvest",
    "Nakhla",
    "Oasis Choice",
    "Arabian Select",
    "Golden Basket",
    "Sahara Home",
    "Rimal",
    "Palm Valley",
    "Al Waha",
    "Eastern Choice",
    "Northern Star"
]
```

### Create Product IDs

```python
product_ids = [
    f"P{i:04d}"
    for i in range(1, 501)
]
```

```python
product_ids[:10]
```

### Generate Category/Subcategory

```python
category_weights = {
    "Food & Beverages": 0.22,
    "Personal Care": 0.12,
    "Household": 0.13,
    "Beauty": 0.10,
    "Health & Wellness": 0.10,
    "Baby Care": 0.08,
    "Electronics": 0.10,
    "Fresh Food": 0.15
}
```

```python
category_list = list(category_weights.keys())
category_probabilities = list(category_weights.values())
```

### Generate categories

```python
product_categories = np.random.choice(
    category_list,
    size=500,
    p=category_probabilities
)
```

### Generate subcategory

```python
product_categories = np.random.choice(
    category_list,
    size=500,
    p=category_probabilities
)
```

### Generate Brands

```python
product_brands = np.random.choice(
    brands,
    size=500
)
```

### Category Product Names

```python
product_name = [
    f"{brand} {subcategory} Item {i:03d}"
    for i, (brand, subcategory) 
    in enumerate(
        zip(product_brands, product_categories),
        start=1
    )
]
```

### Generate Product Costs

```python
cost_ranges = {
    "Food & Beverages": (2, 80),
    "Personal Care": (5, 120),
    "Household": (5, 150),
    "Beauty": (10, 250),
    "Health & Wellness": (10, 200),
    "Baby Care": (8, 180),
    "Electronics": (20, 800),
    "Fresh Food": (3, 100)
}
```

```python
unit_costs = [
    round(
        np.random.uniform(
            cost_ranges[category][0],
            cost_ranges[category][1]
        ),
        2
    )
    for category in product_categories
]
```

### Generate Selling Prices

```python
markup_ranges = {
    "Food & Beverages": (0.15, 0.35),
    "Personal Care": (0.25, 0.50),
    "Household": (0.20, 0.40),
    "Beauty": (0.35, 0.70),
    "Health & Wellness": (0.30, 0.60),
    "Baby Care": (0.20, 0.45),
    "Electronics": (0.10, 0.30),
    "Fresh Food": (0.20, 0.40)
}
```

```python
selling_prices = [
    round(
        cost * (
            1 + np.random.uniform(
                markup_ranges[category][0],
                markup_ranges[category][1]
            )
        ),
        2
    )
    for cost, category in zip(
        unit_costs,
        product_categories
    )
]
```

### Create the DataFrame

```python
df_product = pd.DataFrame({
    "product_id": product_ids,
    "product_name": product_name,
    "category": product_categories,
    "subcategory": product_categories,
    "brand": product_brands,
    "unit_cost": unit_costs,
    "selling_price": selling_prices
})
```
